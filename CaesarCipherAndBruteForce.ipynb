```python
# Part A: Caesar Cipher Encryption

# Question 1: Accept plaintext and key as input
plaintext = input("Enter plaintext: ")
key_input = input("Enter key: ")


# Question 2: Validate the inputs

# Validate plaintext
if not plaintext.isascii() or not plaintext.isalpha():
    print("Error: Plaintext must contain only A-Z or a-z.")

else:

    # Validate that the key is an integer
    try:
        key = int(key_input)

        # Validate that the key is between 0 and 25
        if key < 0 or key > 25:
            print("Error: Key must be an integer between 0 and 25.")

        else:

            # Question 4: Generate ciphertext
            ciphertext = ""

            for character in plaintext:

                # Question 5: Preserve uppercase letters
                if character.isupper():
                    shifted = (ord(character) - ord('A') + key) % 26 + ord('A')
                    ciphertext += chr(shifted)

                # Preserve lowercase letters
                else:
                    shifted = (ord(character) - ord('a') + key) % 26 + ord('a')
                    ciphertext += chr(shifted)

            # Display the ciphertext
            print("Ciphertext:", ciphertext)

    except ValueError:
        # Question 3: Display error for invalid key
        print("Error: Key must be an integer.")
```

```python

# Part 2: Caesar Cipher Decryption

# Question 1: Accept ciphertext and key as input
ciphertext = input("Enter ciphertext: ")
key_input = input("Enter key: ")


# Question 2: Validate the key

try:
    key = int(key_input)

    if key < 0 or key > 25:
        print("Error: Key must be an integer between 0 and 25.")

    else:

        # Question 3: Apply Caesar Cipher decryption
        plaintext = ""

        for character in ciphertext:

            # Decrypt uppercase letters
            if character.isupper():
                shifted = (ord(character) - ord('A') - key) % 26 + ord('A')
                plaintext += chr(shifted)

            # Decrypt lowercase letters
            else:
                shifted = (ord(character) - ord('a') - key) % 26 + ord('a')
                plaintext += chr(shifted)

        # Question 4: Display recovered plaintext
        print("Recovered plaintext:", plaintext)


except ValueError:
    print("Error: Key must be an integer.")
```

```python
# Part C: Cryptanalysis - Brute Force Attack

from docx import Document


# Question 2: Read the ciphertext from the Word document
document = Document("Ciphertext.docx")

ciphertext = ""

for paragraph in document.paragraphs:
    ciphertext += paragraph.text + "\n"


# Create/clear the output file
with open("output.txt", "w") as file:
    file.write("Caesar Cipher Brute Force Results\n")
    file.write("=" * 50 + "\n\n")


# Question 1: Try all possible keys from 0 to 25
for key in range(26):

    plaintext = ""

    # Generate the decrypted plaintext
    for character in ciphertext:

        # Decrypt uppercase letters
        if character.isupper():
            shifted = (ord(character) - ord('A') - key) % 26 + ord('A')
            plaintext += chr(shifted)

        # Decrypt lowercase letters
        elif character.islower():
            shifted = (ord(character) - ord('a') - key) % 26 + ord('a')
            plaintext += chr(shifted)

        # Keep spaces, punctuation and line breaks unchanged
        else:
            plaintext += character


    # Question 3: Write the decrypted text and key to the output file
    with open("output.txt", "a") as file:
        file.write("Key " + str(key) + ":\n")
        file.write(plaintext)
        file.write("\n")
        file.write("-" * 50 + "\n")
```
